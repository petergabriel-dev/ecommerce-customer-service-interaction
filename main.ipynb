## Possible Findings with the Given Data
- Count the number of orders fulfilled by each agent.
- Count the number of orders fulfilled by agents under a certain manager.
- Count the number of orders fulfilled by supervisors.

```python
df.columns

# 1) Count orders/tickets per agent
# If you have a unique order/ticket id use it; otherwise count rows per agent.
order_col = 'Order_id' if 'Order_id' in df.columns else None

if order_col:
    per_agent = df.groupby('Agent_name')[order_col].nunique().sort_values(ascending=False)
else:
    per_agent = df['Agent_name'].value_counts().sort_values(ascending=False)

per_agent

# 2) Count orders per manager (and manager -> agent breakdown) if Manager exists
if 'Manager' in df.columns:
    per_manager = df.groupby('Manager').size().sort_values(ascending=False)
    per_manager_agent = df.groupby(['Manager', 'Agent_name']).size().sort_values(ascending=False)
    print(per_manager)
    print(per_manager_agent)

# 3) Count orders per supervisor if Supervisor exists
if 'Supervisor' in df.columns:
    per_supervisor = df.groupby('Supervisor').size().sort_values(ascending=False)
    per_supervisor

# 4) Save summaries and plot a quick chart
per_agent.head(20).to_csv('per_agent_counts.csv')
per_agent.head(20).plot(kind='barh', figsize=(8,6))
plt.xlabel('Count')
plt.title('Top agents by ticket/order count')
plt.tight_layout()
```

```python

# Supervisor summaries and breakdowns
if 'Supervisor' in df.columns:
    # use unique order id if available to avoid double counting
    if 'Order_id' in df.columns:
        per_supervisor = df.groupby('Supervisor')['Order_id'].nunique().sort_values(ascending=False)
        per_supervisor_agent = df.groupby(['Supervisor', 'Agent_name'])['Order_id'].nunique().sort_values(ascending=False)
    else:
        per_supervisor = df['Supervisor'].value_counts().sort_values(ascending=False)
        per_supervisor_agent = df.groupby(['Supervisor', 'Agent_name']).size().sort_values(ascending=False)

    # show & save top supervisors
    print("Top supervisors (by orders/tickets):")
    print(per_supervisor.head(20))
    per_supervisor.head(20).to_csv('per_supervisor_counts.csv')

    # plot
    per_supervisor.head(20).plot(kind='barh', figsize=(8,6))
    plt.xlabel('Count')
    plt.title('Top supervisors by ticket/order count')
    plt.tight_layout()
    plt.show()

    # breakdown supervisor -> agent
    print("\nSupervisor -> Agent breakdown (top rows):")
    print(per_supervisor_agent.head(40))

    # optional: manager -> supervisor if Manager exists
    if 'Manager' in df.columns:
        if 'Order_id' in df.columns:
            per_manager_supervisor = df.groupby(['Manager', 'Supervisor'])['Order_id'].nunique().sort_values(ascending=False)
        else:
            per_manager_supervisor = df.groupby(['Manager', 'Supervisor']).size().sort_values(ascending=False)
        print("\nManager -> Supervisor counts (top rows):")
        print(per_manager_supervisor.head(40))
else:
    print("No 'Supervisor' column found in df.")
```
